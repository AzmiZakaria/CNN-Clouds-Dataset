```python
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

```

```python
img_height, img_width = 32, 32
batch_size = 20
```

#### 1 Chargement du train + validation à partir du même dossier

```python

train_dataset = tf.keras.utils.image_dataset_from_directory(
    "./clouds_train",
    image_size=(img_height, img_width),
    batch_size=batch_size
)
```

```python
test_dataset = tf.keras.utils.image_dataset_from_directory(
    "./clouds_test",
    validation_split=0.2,  
    subset="training",
    seed=123,  
    image_size=(img_height, img_width),
    batch_size=batch_size
)
```

```python
validation_dataset = tf.keras.utils.image_dataset_from_directory(
    "./clouds_test",
    validation_split=0.2,
    subset="validation",
    seed=123,
    image_size=(img_height, img_width),
    batch_size=batch_size
)
```

### 3 Visualisation

```python
class_names = train_dataset.class_names
plt.figure(figsize=(10,10))
for images, labels in train_dataset.take(1):
    for i in range(9):
        plt.subplot(5,5,i+1)
        plt.imshow(images[i].numpy().astype("uint8"))
        plt.title(class_names[labels[i]])
        plt.axis("off")
plt.show()
```

### 4 Modèle CNN

```python
model = tf.keras.models.Sequential([
    tf.keras.layers.Rescaling(1./255, input_shape=(img_width, img_height, 3)),

    tf.keras.layers.Conv2D(32, (3,3), activation='relu', padding='same'),
    tf.keras.layers.BatchNormalization(),
    tf.keras.layers.MaxPooling2D(2,2),

    tf.keras.layers.Conv2D(64, (3,3), activation='relu', padding='same'),
    tf.keras.layers.BatchNormalization(),
    tf.keras.layers.MaxPooling2D(2,2),

    tf.keras.layers.Conv2D(128, (3,3), activation='relu', padding='same'),
    tf.keras.layers.BatchNormalization(),
    tf.keras.layers.MaxPooling2D(2,2),

    tf.keras.layers.Flatten(),
    tf.keras.layers.Dense(128, activation='relu'),
    tf.keras.layers.Dropout(0.4),
    tf.keras.layers.Dense(64, activation='relu'),
    tf.keras.layers.Dropout(0.3),

    tf.keras.layers.Dense(len(class_names), activation='softmax')
])
```

### 5 Compilation

```python
optimizer = tf.keras.optimizers.Adam(learning_rate=0.0005)
model.compile(
    optimizer=optimizer,
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

# callbacks

```python
reduce_lr = tf.keras.callbacks.ReduceLROnPlateau(monitor='val_loss', factor=0.2, patience=3, min_lr=0.0001)
early_stopping = tf.keras.callbacks.EarlyStopping(monitor='val_loss', patience=5, restore_best_weights=True)
```

### 6 Entraînement

```python
model.fit(train_dataset, validation_data=validation_dataset, epochs=60,callbacks=[reduce_lr])
# model.fit(train_dataset, validation_data=validation_dataset, epochs=40)
```

```python
model.evaluate(test_dataset, verbose=2)
# 25/25 - 0s - 14ms/step - accuracy: 0.6728 - loss: 0.8375    |     epoch =50
# 25/25 - 0s - 9ms/step - accuracy: 0.7181 - loss: 0.7704     |     epoch=40
# 31/31 - 0s - 14ms/step - accuracy: 0.6934 - loss: 0.8061    |     epoch=40, batch_size=16
# 16/16 - 2s - 123ms/step - accuracy: 0.6358 - loss: 1.3306   |     epoch=40, batch_size=32, img_size=128*128
# 16/16 - 2s - 96ms/step - accuracy: 0.6564 - loss: 0.9530    |     epoch=40, batch_size=32, img_size=128*128 , callbacks=[reduce_lr, early_stopping]
# 20/20 - 0s - 10ms/step - accuracy: 0.7147 - loss: 0.7337    |     epoch=60, batch_size=20, img_size=32*32 , callbacks=[reduce_lr, early_stopping], sequential()
# 20/20 - 0s - 18ms/step - accuracy: 0.8098 - loss: 0.7650    |     same + min_learning_rate=0.0001
# 20/20 - 0s - 16ms/step - accuracy: 0.6761 - loss: 0.8110    |     same + min_learning_rate=1e-6
# 
```

```python

plt.figure(figsize=(10, 10))
for images, labels in test_dataset.take(1):  # take one batch
    for i in range(9):  # show 9 images
        ax = plt.subplot(3, 3, i + 1)
        plt.imshow(images[i].numpy().astype("uint8"))
        plt.title(class_names[labels[i]])
        plt.axis("off")
plt.show()
```

```python
for images,labels in test_dataset.take(1):
    predictions = model.predict(images)
    predicted_classes=np.argmax(predictions,axis=1)
    true_classes=labels.numpy()
    
for i in range(len(images)):
    print("Predicted: {}, Actual: {}".format(class_names[predicted_classes[i]],class_names[true_classes[i]]))
```

```python
from sklearn.metrics import confusion_matrix, classification_report
```

```python
# --- Confusion Matrix ---
cm = confusion_matrix(true_classes, predicted_classes)
print("\nConfusion Matrix:\n", cm)
```

```python
# Plot confusion matrix
plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=class_names,
            yticklabels=class_names)
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.title('Confusion Matrix')
plt.show()

```
